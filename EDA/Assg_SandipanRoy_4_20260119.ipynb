---

*   The feature- 'Fare' is continuous in nature
*   Probability of a continuous data is zero, however we can find the probability of a continuous variable over a given range.
*   Since our earlier approach shown us that w.r.t. 'Ticket', 'Cabin' or subset of 'Name' is not enough to find the 'Embarked' info for the missing data.
*   Hence, we are creating a mask where we will mention the 'Fare' over a range,'Embarked' with no missing data and 'Survived' equals to True or 1.
*   The objective is to get a close approximation by using the central tendency and fill the missing data w.r.t. the feature 'Fare' and 'Survived'




```python
embarkment_mask = df_clean.query("Fare <= 80.00000 & Fare >=70.00000 & ~Embarked.isna() & Survived == 1")
embarkment_mask.info()
```

```python
embarkment_mask
```

```python
embarkment_mask["Embarked"].mode()
```

```python
df_clean["Embarked"].fillna(value=embarkment_mask["Embarked"].mode()[0], inplace=True)
df_clean.info()
```
