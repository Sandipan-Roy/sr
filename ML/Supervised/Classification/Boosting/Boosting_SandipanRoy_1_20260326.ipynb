```python
def acquire_data(url, sourceType="csv", verbose=False):
    """
    This function acquires data from the given URL and returns a DataFrame.

    Parameters:
    url (str): The URL from which to acquire the data.
    sourceType (str): The type of the data source (default is "csv").
    verbose (bool): If True, prints the shape of the DataFrame and its columns.

    Returns:
    pd.DataFrame: The acquired data as a DataFrame.
    """

    if sourceType.lower() != "csv":
        raise ValueError("Currently, only 'csv' source type is supported.")

    # Reading the CSV file from the provided URL
    df = pd.read_csv(url)

    # If verbose is True, print the shape of the DataFrame and its columns
    if verbose:
        print(f"DataFrame Shape: {df.shape}, i.e., {df.shape[0]} rows and {df.shape[1]} columns.")
        print(f"DataFrame Columns: {df.columns.tolist()}")

    return df
```

```python
def analyze_data(df):
    """
    This function performs basic analysis on the given DataFrame and prints the results.

    Parameters:
    df (pd.DataFrame): The DataFrame to analyze.

    Returns:
    __dict__: A dictionary containing the results of the numerical columns and categorical columns list.
    """

    # Displaying the first 5 rows of the DataFrame
    print("First 5 rows of the DataFrame:")
    print(df.head())

    # Displaying the last 5 rows of the DataFrame
    print("\nLast 5 rows of the DataFrame:")
    print(df.tail())

    # Displaying the shape of the DataFrame
    print(f"\nShape of the DataFrame: {df.shape}")

    # Displaying the data types of each column
    print("\nData types of each column:")
    print(df.dtypes)

    #Displaying information about the DataFrame
    print("\nInformation about the DataFrame:")
    print(df.info())

    # Displaying the memory usage of the DataFrame
    print(f"\nMemory usage of the DataFrame: {(df.memory_usage(deep=True)).sum() / 1024:.2f} KB")  # Converting bytes to kilobytes
    print(((df.memory_usage(deep=True)) / 1024).round(2))  # Converting bytes to kilobytes

    # Listing numerical columns
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    # Listing categorical columns
    categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
    print("\nWe have {} numerical features: {}".format(len(numerical_cols), numerical_cols))
    print("We have {} categorical features: {}".format(len(categorical_cols), categorical_cols))

    print("\nNumber of duplicated rows: {} ({}% of total rows)".format(
                    df.duplicated().sum(),
                    round(df.duplicated().sum() / df.shape[0] * 100, 2)
                    ))

    if len(numerical_cols) > 0:
      # Displaying summary statistics for numerical columns
      print("\nSummary statistics for numerical columns:")
      print(df.describe())

    if len(categorical_cols) > 0:
      # Displaying summary statistics for categorical columns
      print("\nSummary statistics for categorical columns:")
      print(df.describe(include=['object']))


    # Returning the list of numerical and categorical columns
    return {
        "numerical_columns": numerical_cols if len(numerical_cols) > 0 else None,
        "categorical_columns": categorical_cols if len(categorical_cols) > 0 else None
    }
```

```python
def get_proportion_of_categorical_values(df, column_name):
    """
    This function calculates the proportion of each unique value in a categorical column.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the categorical column.
    column_name (str): The name of the categorical column for which to calculate proportions.

    Returns:
    None (prints proportions for each column).
    """
    not_in_list2 = list(set(column_name) - set(df.columns))

    # Checking if the specified column is in the DataFrame
    if len(not_in_list2) > 0:
        raise ValueError(f"Column(s) '{', '.join(not_in_list2)}' not found in the DataFrame.")

    # Calculating the proportion of each unique value in the specified column
    for col in column_name:
        proportions = df[col].value_counts(normalize=True) * 100  # Multiplying by 100 to get percentages
        print(f"\nProportions for column '{col}' in %:")
        print(proportions)
        print("-" * 50)  # Adding a separator for better readability
```

```python
def copy_dataframe(df):
    """
    This function creates and returns a copy of the given DataFrame
    to ensure that the original DataFrame remains unchanged during analysis.
    Also it removes duplicate data points if there are any in the DataFrame.

    Parameters:
    df (pd.DataFrame): The DataFrame to copy.

    Returns:
    pd.DataFrame: A copy of the input DataFrame.
    """
    df_copy = df.copy()

    # dropping duplicate data and keeping the first occurrence of each duplicate
    df_copy.drop_duplicates(inplace=True, keep='first')
    return df_copy
```

```python
def plot_univariate_distribution(df, column, plot_type="histogram", plot_width=12, plot_height=4):
    """
    This function plots the univariate distribution of a specified column in the DataFrame 
    using the specified plot type (histogram, boxplot, or countplot). It also calculates 
    and prints the Interquartile Range (IQR) of the specified column.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the data.
    column (str): The name of the column to plot.
    plot_type (str): The type of plot to create ("histogram" or "boxplot" or "countplot").

    Returns:
    None
    """

    fig, (ax1,ax2) = plt.subplots(nrows=1, ncols=2, figsize=(plot_width, plot_height))

    sns.kdeplot(df[column], shade=True, color='blue', ax=ax1)
    ax1.set_title(f'Distribution of {column.capitalize()}')
    ax1.set_xlabel(column.capitalize())
    #plt.figure(figsize=(6, 4))
    
    #plt.title(f'Distribution of {column.capitalize()}')
    #plt.xlabel(column.capitalize())

    if plot_type != "boxplot":
        if plot_type == "histogram":
            ax2 = sns.histplot(df[column], kde=True, color='blue')
            ax2.set_ylabel('Frequency')
            ax2.set_title(f'{plot_type.capitalize()} of {column.capitalize()}')
            #plt.ylabel('Frequency')
            #plt.title(f'{plot_type.capitalize()} of {column.capitalize()}')
        elif plot_type == "countplot":
            ax2 = sns.countplot(x=df[column], palette='magma')
            ax2.set_ylabel('Count')
            ax2.set_title(f'{plot_type.capitalize()} of {column.capitalize()}')
        else:
            raise ValueError("Invalid plot type. Use 'histogram' or 'boxplot' or 'countplot'.")

        for p in ax2.patches:
            percentage = '{:.2f}%'.format(100 * p.get_height() / total_datapoints)
            x = p.get_x() + p.get_width()/5
            y = p.get_y() + p.get_height()
            ax2.annotate(percentage, (x, y))

    else:
        ax2 = sns.boxplot(x=df[column], palette='magma')
        ax2.set_title(f'{plot_type.capitalize()} of {column.capitalize()}')


    
    ax2.set_xlabel(column.capitalize())
    plt.show()

    iqr_column = df[column].quantile(0.75) - df[column].quantile(0.25)
    print(f" IQR of {column.capitalize()}: {iqr_column}")
```

```python
def plot_bivariate_relationship(df, x_column, y_column, x_label=None, y_label=None, plot_type="scatter", plot_width=6, plot_height=4):
    """
    This function plots the bivariate relationship between two specified columns in the DataFrame.

    Parameters:
    df (pd.DataFrame): The DataFrame containing the data.
    x_column (str): The name of the column to be plotted on the x-axis.
    y_column (str): The name of the column to be plotted on the y-axis.
    x_label (str): The label for the x-axis.
    y_label (str): The label for the y-axis.
    plot_type (str): The type of plot to create ("scatter" or "line" or "crosstab" or "boxplot").
    plot_width (int): The width of the plot in inches.
    plot_height (int): The height of the plot in inches.

    Returns:
    None
    """

    fig = plt.figure(figsize=(plot_width, plot_height))

    if plot_type == "scatter":
        ax = sns.scatterplot(x=df[x_column], y=df[y_column], color='blue')

        plt.title(f'Scatter Plot of {y_column.capitalize() if y_label is None else y_label.capitalize()} vs {x_column.capitalize() if x_label is None else x_label.capitalize()}')

    elif plot_type == "line":
        ax = sns.lineplot(x=df[x_column], y=df[y_column], color='blue')

        plt.title(f'Line Plot of {y_column.capitalize() if y_label is None else y_label.capitalize()} vs {x_column.capitalize() if x_label is None else x_label.capitalize()}')

    elif plot_type == "countplot":
        ax = sns.countplot(x=df[x_column], hue=df[y_column], palette='magma')

        for p in ax.patches:
            percentage = '{:.2f}%'.format(100 * p.get_height() / total_datapoints)
            x = p.get_x() + p.get_width()/5
            y = p.get_y() + p.get_height()
            ax.annotate(percentage, (x, y))

        plt.title(f'Count Plot of {y_column.capitalize() if y_label is None else y_label.capitalize()} vs {x_column.capitalize() if x_label is None else x_label.capitalize()}')
    elif plot_type == "crosstab":
        crosstab = pd.crosstab(df[x_column], df[y_column])
        ax = crosstab.plot(kind='bar', stacked=True, figsize=fig)

        for p in ax.patches:
            percentage = '{:.2f}%'.format(100 * p.get_height() / total_datapoints)
            x = p.get_x() + p.get_width()/5
            y = p.get_y() + p.get_height()
            ax.annotate(percentage, (x, y))

        plt.title(f'Cross-tabulation of {y_column.capitalize() if y_label is None else y_label.capitalize()} vs {x_column.capitalize() if x_label is None else x_label.capitalize()}')

    elif plot_type == "boxplot":
        ax = sns.boxplot(x=df[x_column], y=df[y_column])

        plt.title(f'Box Plot of {y_column.capitalize() if y_label is None else y_label.capitalize()} vs {x_column.capitalize() if x_label is None else x_label.capitalize()}')

    else:
        raise ValueError("Invalid plot type. Use 'scatter' or 'line'  or 'crosstab' or 'boxplot'.")


    plt.ylabel(y_column.capitalize() if y_label is None else y_label.capitalize())
    plt.xlabel(x_column.capitalize() if x_label is None else x_label.capitalize())

    plt.grid(visible=True)
    plt.show()
```

```python
def winsorize_iqr_by_class(df, group_col='is_safe', cols=None, k=1.5):
    """
    This function performs winsorization on the specified columns of a DataFrame based on the 
    Interquartile Range (IQR) for each class in a specified grouping column.

    Parameters:
    df (pd.DataFrame): The DataFrame to be processed.
    group_col (str): The name of the column to group by (default is 'is_safe').
    cols (list): The list of columns to be winsorized (default is None, which means all columns).
    k (float): The multiplier for the IQR (default is 1.5).

    Returns:
    pd.DataFrame: A DataFrame with the specified columns winsorized based on the IQR for each class.
    """

    if cols is None:
        cols = df.columns.tolist()

    bounds =pd.DataFrame(columns=["Column", "Lower Bound", "Upper Bound"])

    for col in cols:
        # Grouping the DataFrame by the specified grouping column and calculating the IQR for each group
        grouped = df.groupby(group_col)[col]
        Q1 = grouped.transform(lambda x: x.quantile(0.25))
        Q3 = grouped.transform(lambda x: x.quantile(0.75))
        IQR = Q3 - Q1

        # Calculating the lower and upper bounds for winsorization
        lower_bound = Q1 - k * IQR
        upper_bound = Q3 + k * IQR

        sub_bounds = pd.DataFrame({
            'Column': col,
            'Lower Bound': lower_bound,
            'Upper Bound': upper_bound
        })
        
        if len(bounds) == 0:
            bounds = sub_bounds
        else:
            if col not in bounds['Column'].values:
                bounds = pd.concat([bounds, sub_bounds], ignore_index=True)

        # Applying winsorization to the specified column based on the calculated bounds
        df[col] = np.where(df[col] < lower_bound, lower_bound, np.where(df[col] > upper_bound, upper_bound, df[col]))

    return df, bounds.drop_duplicates(keep='first')
```

```python
def compare_plots_before_after_winsorization(df_before, df_after, column, target_class, plot_width=12, plot_height=4):
    """
    This function compares the distribution of a specified column before and after winsorization 
    by plotting the distributions side by side.

    Parameters:
    df_before (pd.DataFrame): The DataFrame before winsorization.
    df_after (pd.DataFrame): The DataFrame after winsorization.
    column (str): The name of the column to compare.
    plot_width (int): The width of the plot in inches (default is 12).
    plot_height (int): The height of the plot in inches (default is 4).

    Returns:
    None
    """

    fig1, axes1 = plt.subplots(nrows=1, ncols=2, figsize=(plot_width, plot_height))

    sns.kdeplot(df_before[column], shade=True, color='blue', ax=axes1[0])
    axes1[0].set_title(f'Distribution of {column.capitalize()} Before Winsorization')
    axes1[0].set_xlabel(column.capitalize())

    sns.kdeplot(df_after[column], shade=True, color='orange', ax=axes1[1])
    axes1[1].set_title(f'Distribution of {column.capitalize()} After Winsorization')
    axes1[1].set_xlabel(column.capitalize())

    plt.tight_layout()
    plt.show()

    fig2, axes2 = plt.subplots(nrows=1, ncols=2, figsize=(plot_width, plot_height))

    sns.boxplot(x=df_before[target_class], y=df_before[column],width=0.9,palette='pastel', ax=axes2[0])
    axes2[0].set_title(f'{column.capitalize()} by {target_class.capitalize()} Before Winsorization')
    axes2[0].set_xlabel(target_class.capitalize())
    axes2[0].set_ylabel(column.capitalize())

    sns.boxplot(x=df_after[target_class], y=df_after[column],width=0.9,palette='pastel', ax=axes2[1])
    axes2[1].set_title(f'{column.capitalize()} by {target_class.capitalize()} After Winsorization')
    axes2[1].set_xlabel(target_class.capitalize())
    axes2[1].set_ylabel(column.capitalize())

    plt.tight_layout()
    plt.show()
```

```python
# ========================================
# COMPREHENSIVE OPTIMAL THRESHOLD ANALYSIS
# ========================================
# This cell demonstrates 4 methods to find optimal thresholds for all 3 models

from sklearn.metrics import precision_recall_curve, f1_score
import numpy as np

print("\n" + "="*80)
print("OPTIMAL THRESHOLD COMPARISON FOR ALL 3 BOOSTING MODELS")
print("="*80)

threshold_results = {}

for model_name, model in models.items():
    print(f"\n{'='*80}")
    print(f"MODEL: {model_name.upper()}")
    print(f"{'='*80}")
    
    # Get probabilities for validation set
    y_val_proba = model.predict_proba(X_val_scaled)[:, 1]
    
    # ========== METHOD 1: Youden's J Statistic (ROC-based) ==========
    from sklearn.metrics import roc_curve, roc_auc_score
    fpr, tpr, thresholds_roc = roc_curve(y_val, y_val_proba)
    j_scores = tpr - fpr
    optimal_idx_youden = np.argmax(j_scores)
    threshold_youden = thresholds_roc[optimal_idx_youden]
    roc_auc = roc_auc_score(y_val, y_val_proba)
    
    # ========== METHOD 2: Precision-Recall F1 Optimization ==========
    precision, recall, thresholds_pr = precision_recall_curve(y_val, y_val_proba)
    f1_scores = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-10)
    optimal_idx_f1 = np.argmax(f1_scores)
    threshold_f1 = thresholds_pr[optimal_idx_f1]
    
    # ========== METHOD 3: Balanced Accuracy ==========
    from sklearn.metrics import balanced_accuracy_score
    # Test multiple thresholds to find best balanced accuracy
    test_thresholds = np.linspace(0, 1, 100)
    balanced_accs = []
    for thresh in test_thresholds:
        y_pred_thresh = (y_val_proba >= thresh).astype(int)
        balanced_acc = balanced_accuracy_score(y_val, y_pred_thresh)
        balanced_accs.append(balanced_acc)
    optimal_idx_balanced = np.argmax(balanced_accs)
    threshold_balanced = test_thresholds[optimal_idx_balanced]
    
    # ========== METHOD 4: Default Threshold (0.5) for Comparison ==========
    threshold_default = 0.5
    
    print(f"\nROC-AUC Score: {roc_auc:.4f}")
    print(f"\n{'METHOD':<30} {'THRESHOLD':<12} {'VALIDATION F1':<15}")
    print("-"*60)
    
    # Evaluate all thresholds on validation set
    for method_name, threshold in [
        ("1. Youden's J Statistic", threshold_youden),
        ("2. Precision-Recall F1", threshold_f1),
        ("3. Balanced Accuracy", threshold_balanced),
        ("4. Default (0.5)", threshold_default)
    ]:
        y_pred_thresh = (y_val_proba >= threshold).astype(int)
        val_f1 = f1_score(y_val, y_pred_thresh)
        print(f"{method_name:<30} {threshold:<12.4f} {val_f1:<15.4f}")
    
    # Store results
    threshold_results[model_name] = {
        'youden': threshold_youden,
        'f1': threshold_f1,
        'balanced_acc': threshold_balanced,
        'default': threshold_default,
        'roc_auc': roc_auc
    }
    
    # ========== PERFORMANCE AT OPTIMAL THRESHOLDS ==========
    print(f"\n{'DETAILED METRICS AT OPTIMAL THRESHOLDS (Validation Set)'}")
    print("-"*80)
    
    for method_name, threshold in [
        ("Youden's J Statistic", threshold_youden),
        ("Precision-Recall F1", threshold_f1),
        ("Balanced Accuracy", threshold_balanced)
    ]:
        print(f"\n{method_name}: Threshold = {threshold:.4f}")
        y_pred_thresh = (y_val_proba >= threshold).astype(int)
        
        from sklearn.metrics import precision_score, recall_score
        acc = accuracy_score(y_val, y_pred_thresh)
        prec = precision_score(y_val, y_pred_thresh, zero_division=0)
        rec = recall_score(y_val, y_pred_thresh, zero_division=0)
        f1 = f1_score(y_val, y_pred_thresh)
        bal_acc = balanced_accuracy_score(y_val, y_pred_thresh)
        
        print(f"  Accuracy: {acc:.4f} | Precision: {prec:.4f} | Recall: {rec:.4f}")
        print(f"  F1-Score: {f1:.4f} | Balanced Accuracy: {bal_acc:.4f}")
        print(f"  Confusion Matrix: {confusion_matrix(y_val, y_pred_thresh).ravel()}")
    
    print("\n")

print("\n" + "="*80)
print("RECOMMENDATION:")
print("="*80)
print("""
For this brain stroke dataset (imbalanced, medical context):
- USE YOUDEN'S J STATISTIC: Best for ROC curve optimization
- USE PRECISION-RECALL F1: Best for balanced precision-recall trade-off
- AVOID DEFAULT 0.5: Often biased towards majority class

For production deployment, choose based on:
- HIGH RECALL NEEDED: Use lower threshold (catch more strokes, false alarms acceptable)
- HIGH PRECISION NEEDED: Use higher threshold (minimize false alarms)
- BALANCED: Use Youden's or F1-based threshold
""")
```

```python
# ========================================
# COMPREHENSIVE MODEL COMPARISON - ALL 9 MODELS SIDE-BY-SIDE
# ========================================

import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, precision_score, recall_score

print("\n" + "="*140)
print("COMPREHENSIVE MODEL COMPARISON: ORIGINAL vs REGULARIZED")
print("="*140)

# Create comparison dataframe
comparison_data = []

# Iterate through both original and improved models
for comparison_type, models_dict in [("ORIGINAL", models), ("REGULARIZED", improved_models)]:
    
    for model_name, model in models_dict.items():
        
        # Get predictions and probabilities for all sets
        y_train_pred = model.predict(X_train_scaled)
        y_val_pred = model.predict(X_val_scaled)
        y_test_pred = model.predict(X_test_scaled)
        
        y_train_proba = model.predict_proba(X_train_scaled)[:, 1]
        y_val_proba = model.predict_proba(X_val_scaled)[:, 1]
        y_test_proba = model.predict_proba(X_test_scaled)[:, 1]
        
        # Calculate metrics for TRAINING SET
        train_acc = accuracy_score(y_train_resampled, y_train_pred)
        train_prec = precision_score(y_train_resampled, y_train_pred, zero_division=0)
        train_rec = recall_score(y_train_resampled, y_train_pred, zero_division=0)
        train_f1 = f1_score(y_train_resampled, y_train_pred)
        train_roc = roc_auc_score(y_train_resampled, y_train_proba)
        
        # Calculate metrics for VALIDATION SET
        val_acc = accuracy_score(y_val, y_val_pred)
        val_prec = precision_score(y_val, y_val_pred, zero_division=0)
        val_rec = recall_score(y_val, y_val_pred, zero_division=0)
        val_f1 = f1_score(y_val, y_val_pred)
        val_roc = roc_auc_score(y_val, y_val_proba)
        
        # Calculate metrics for TEST SET
        test_acc = accuracy_score(y_test, y_test_pred)
        test_prec = precision_score(y_test, y_test_pred, zero_division=0)
        test_rec = recall_score(y_test, y_test_pred, zero_division=0)
        test_f1 = f1_score(y_test, y_test_pred)
        test_roc = roc_auc_score(y_test, y_test_proba)
        
        # Calculate generalization gaps
        acc_gap = train_acc - val_acc
        f1_gap = train_f1 - val_f1
        roc_gap = train_roc - val_roc
        
        # Store in list
        comparison_data.append({
            'Model Type': comparison_type,
            'Algorithm': model_name,
            'Train Acc': train_acc,
            'Val Acc': val_acc,
            'Test Acc': test_acc,
            'Train F1': train_f1,
            'Val F1': val_f1,
            'Test F1': test_f1,
            'Train ROC': train_roc,
            'Val ROC': val_roc,
            'Test ROC': test_roc,
            'Acc Gap': acc_gap,
            'F1 Gap': f1_gap,
            'ROC Gap': roc_gap
        })

# Create DataFrame
comparison_df = pd.DataFrame(comparison_data)

# Display full comparison
print("\n" + "="*140)
print("1. ACCURACY METRICS")
print("="*140)
print(comparison_df[['Model Type', 'Algorithm', 'Train Acc', 'Val Acc', 'Test Acc', 'Acc Gap']].to_string(index=False))

print("\n" + "="*140)
print("2. F1-SCORE METRICS (Preferred for Imbalanced Data)")
print("="*140)
print(comparison_df[['Model Type', 'Algorithm', 'Train F1', 'Val F1', 'Test F1', 'F1 Gap']].to_string(index=False))

print("\n" + "="*140)
print("3. ROC-AUC METRICS (Preferred for Classification)")
print("="*140)
print(comparison_df[['Model Type', 'Algorithm', 'Train ROC', 'Val ROC', 'Test ROC', 'ROC Gap']].to_string(index=False))

# ============ RANKING & RECOMMENDATIONS ============
print("\n" + "="*140)
print("MODEL RANKING BY TEST SET PERFORMANCE")
print("="*140)

print("\n📊 RANKED BY TEST F1-SCORE (Best Metric for Imbalanced Data):")
print("-"*140)
f1_rank = comparison_df.nlargest(9, 'Test F1')[['Model Type', 'Algorithm', 'Test F1', 'Test ROC', 'Test Acc', 'F1 Gap']]
for idx, (i, row) in enumerate(f1_rank.iterrows(), 1):
    gap_indicator = "✓ GOOD" if row['F1 Gap'] < 0.05 else "⚠️ WARNING"
    print(f"{idx}. {row['Algorithm']:<20} ({row['Model Type']:<12}) | Test F1: {row['Test F1']:.4f} | Test ROC: {row['Test ROC']:.4f} | Generalization: {gap_indicator} (Gap: {row['F1 Gap']:.4f})")

print("\n📊 RANKED BY TEST ROC-AUC:")
print("-"*140)
roc_rank = comparison_df.nlargest(9, 'Test ROC')[['Model Type', 'Algorithm', 'Test ROC', 'Test F1', 'Test Acc', 'ROC Gap']]
for idx, (i, row) in enumerate(roc_rank.iterrows(), 1):
    gap_indicator = "✓ GOOD" if row['ROC Gap'] < 0.05 else "⚠️ WARNING"
    print(f"{idx}. {row['Algorithm']:<20} ({row['Model Type']:<12}) | Test ROC: {row['Test ROC']:.4f} | Test F1: {row['Test F1']:.4f} | Generalization: {gap_indicator} (Gap: {row['ROC Gap']:.4f})")

# ============ BEST MODEL ANALYSIS ============
print("\n" + "="*140)
print("BEST MODEL SELECTION")
print("="*140)

best_f1_idx = comparison_df['Test F1'].idxmax()
best_roc_idx = comparison_df['Test ROC'].idxmax()
best_generalization_idx = comparison_df['F1 Gap'].idxmin()

best_f1_model = comparison_df.loc[best_f1_idx]
best_roc_model = comparison_df.loc[best_roc_idx]
best_gen_model = comparison_df.loc[best_generalization_idx]

print(f"\n🥇 BEST BY TEST F1-SCORE:")
print(f"   Model: {best_f1_model['Algorithm']} ({best_f1_model['Model Type']})")
print(f"   Test F1: {best_f1_model['Test F1']:.4f} | Test ROC: {best_f1_model['Test ROC']:.4f} | Test Acc: {best_f1_model['Test Acc']:.4f}")
print(f"   Generalization Gap: {best_f1_model['F1 Gap']:.4f}")

print(f"\n🥈 BEST BY TEST ROC-AUC:")
print(f"   Model: {best_roc_model['Algorithm']} ({best_roc_model['Model Type']})")
print(f"   Test ROC: {best_roc_model['Test ROC']:.4f} | Test F1: {best_roc_model['Test F1']:.4f} | Test Acc: {best_roc_model['Test Acc']:.4f}")
print(f"   Generalization Gap: {best_roc_model['ROC Gap']:.4f}")

print(f"\n🥉 BEST GENERALIZATION (Lowest Overfitting):")
print(f"   Model: {best_gen_model['Algorithm']} ({best_gen_model['Model Type']})")
print(f"   F1 Gap: {best_gen_model['F1 Gap']:.4f} | Test F1: {best_gen_model['Test F1']:.4f} | Test ROC: {best_gen_model['Test ROC']:.4f}")

# ============ RECOMMENDATION ============
print("\n" + "="*140)
print("FINAL RECOMMENDATIONS")
print("="*140)

if best_gen_model['F1 Gap'] < 0.05:
    print(f"\n✓ RECOMMENDED MODEL: {best_gen_model['Algorithm']} ({best_gen_model['Model Type']})")
    print(f"  Reason: Excellent generalization (F1 Gap: {best_gen_model['F1 Gap']:.4f} < 0.05)")
    print(f"  Performance: Test F1 = {best_gen_model['Test F1']:.4f}, Test ROC = {best_gen_model['Test ROC']:.4f}")
    print(f"  This model balances performance with good generalization to unseen data.")
else:
    print(f"\n✓ RECOMMENDED MODEL: {best_f1_model['Algorithm']} ({best_f1_model['Model Type']})")
    print(f"  Reason: Highest test F1-score = {best_f1_model['Test F1']:.4f}")
    print(f"  Note: Monitor F1 Gap ({best_f1_model['F1 Gap']:.4f}) for potential overfitting")
    print(f"  Alternative with better generalization: {best_gen_model['Algorithm']} (F1={best_gen_model['Test F1']:.4f}, Gap={best_gen_model['F1 Gap']:.4f})")

print("\n" + "="*140)
print("SUMMARY TABLE: ORIGINAL vs REGULARIZED MODELS")
print("="*140)

summary_by_type = comparison_df.groupby('Model Type').agg({
    'Test F1': ['mean', 'max', 'min'],
    'Test ROC': ['mean', 'max', 'min'],
    'F1 Gap': ['mean', 'max'],
}).round(4)

print("\nAverage Metrics by Model Type:")
print(summary_by_type.to_string())

print("\n" + "="*140)
```

```python
# ========================================
# GENERATE PREDICTIONS WITH BEST MODEL + OPTIMAL THRESHOLD
# ========================================

from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score, precision_recall_curve, balanced_accuracy_score
import pandas as pd
import numpy as np

print("\n" + "="*120)
print("BEST MODEL PREDICTION GENERATION WITH OPTIMAL THRESHOLD")
print("="*120)

# Select the best model - prioritize F1-score with good generalization
best_model_name = comparison_df.loc[comparison_df['Test F1'].idxmax(), 'Algorithm']
best_model_type = comparison_df.loc[comparison_df['Test F1'].idxmax(), 'Model Type']

print(f"\n✓ Selected Best Model: {best_model_name} ({best_model_type})")
print(f"  Test F1-Score: {comparison_df.loc[comparison_df['Test F1'].idxmax(), 'Test F1']:.4f}")
print(f"  Test ROC-AUC: {comparison_df.loc[comparison_df['Test F1'].idxmax(), 'Test ROC']:.4f}")
print(f"  Generalization Gap: {comparison_df.loc[comparison_df['Test F1'].idxmax(), 'F1 Gap']:.4f}")

# Get the best model object
if best_model_type == "ORIGINAL":
    best_model = models[best_model_name]
else:  # REGULARIZED
    best_model = improved_models[best_model_name]

# ============ CALCULATE OPTIMAL THRESHOLD ============
print(f"\n{'='*120}")
print(f"CALCULATING OPTIMAL THRESHOLD FOR {best_model_name.upper()}")
print(f"{'='*120}")

# Get probabilities for validation set
y_val_proba_best = best_model.predict_proba(X_val_scaled)[:, 1]

# Method 1: Youden's J Statistic
fpr, tpr, thresholds_roc = roc_curve(y_val, y_val_proba_best)
j_scores = tpr - fpr
optimal_idx_youden = np.argmax(j_scores)
optimal_threshold_youden = thresholds_roc[optimal_idx_youden]

# Method 2: Precision-Recall F1
precision, recall, thresholds_pr = precision_recall_curve(y_val, y_val_proba_best)
f1_scores_thresh = 2 * (precision[:-1] * recall[:-1]) / (precision[:-1] + recall[:-1] + 1e-10)
optimal_idx_f1 = np.argmax(f1_scores_thresh)
optimal_threshold_f1 = thresholds_pr[optimal_idx_f1]

# Method 3: Balanced Accuracy
test_thresholds = np.linspace(0, 1, 100)
balanced_accs = []
for thresh in test_thresholds:
    y_val_pred_thresh = (y_val_proba_best >= thresh).astype(int)
    balanced_acc = balanced_accuracy_score(y_val, y_val_pred_thresh)
    balanced_accs.append(balanced_acc)
optimal_idx_balanced = np.argmax(balanced_accs)
optimal_threshold_balanced = test_thresholds[optimal_idx_balanced]

# Use Youden's J as primary
optimal_threshold = optimal_threshold_youden

print(f"\nOptimal Thresholds Calculated:")
print(f"  Youden's J Statistic:     {optimal_threshold_youden:.4f} (PRIMARY)")
print(f"  Precision-Recall F1:      {optimal_threshold_f1:.4f}")
print(f"  Balanced Accuracy:        {optimal_threshold_balanced:.4f}")
print(f"\n✓ Using Youden's J threshold: {optimal_threshold:.4f}")

# ============ GENERATE PREDICTIONS ON TEST SET ============
print(f"\n{'='*120}")
print(f"GENERATING PREDICTIONS ON TEST SET")
print(f"{'='*120}")

# Get test set probabilities and predictions
y_test_proba_best = best_model.predict_proba(X_test_scaled)[:, 1]
y_test_proba_class0 = best_model.predict_proba(X_test_scaled)[:, 0]

# Predictions with optimal threshold
y_test_pred_optimal = (y_test_proba_best >= optimal_threshold).astype(int)

# Predictions with default threshold (0.5)
y_test_pred_default = best_model.predict(X_test_scaled)

# Confidence scores (max probability between both classes)
confidence_scores = np.maximum(y_test_proba_class0, y_test_proba_best)

# Create comprehensive predictions dataframe
predictions_df = pd.DataFrame({
    'Actual_Class': y_test.values,
    'Probability_Class_0': y_test_proba_class0,
    'Probability_Class_1': y_test_proba_best,
    'Predicted_Class_Default_Threshold': y_test_pred_default,
    'Predicted_Class_Optimal_Threshold': y_test_pred_optimal,
    'Confidence_Score': confidence_scores,
    'Prediction_Correct': y_test_pred_optimal == y_test.values
})

print(f"\nTest Set Size: {len(predictions_df)} samples")
print(f"\nPredictions DataFrame Head (First 10 rows):")
print(predictions_df.head(10).to_string())

# ============ PERFORMANCE METRICS ============
print(f"\n{'='*120}")
print(f"PERFORMANCE METRICS ON TEST SET")
print(f"{'='*120}")

print(f"\nUsing DEFAULT Threshold (0.5):")
print("-"*120)
acc_default = accuracy_score(y_test, y_test_pred_default)
prec_default = precision_score(y_test, y_test_pred_default)
rec_default = recall_score(y_test, y_test_pred_default)
f1_default = f1_score(y_test, y_test_pred_default)
roc_default = roc_auc_score(y_test, y_test_proba_best)

print(f"Accuracy:  {acc_default:.4f}")
print(f"Precision: {prec_default:.4f}")
print(f"Recall:    {rec_default:.4f}")
print(f"F1-Score:  {f1_default:.4f}")
print(f"ROC-AUC:   {roc_default:.4f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_test_pred_default)}")

print(f"\nUsing OPTIMAL Threshold ({optimal_threshold:.4f}):")
print("-"*120)
acc_optimal = accuracy_score(y_test, y_test_pred_optimal)
prec_optimal = precision_score(y_test, y_test_pred_optimal)
rec_optimal = recall_score(y_test, y_test_pred_optimal)
f1_optimal = f1_score(y_test, y_test_pred_optimal)
roc_optimal = roc_auc_score(y_test, y_test_proba_best)

print(f"Accuracy:  {acc_optimal:.4f}")
print(f"Precision: {prec_optimal:.4f}")
print(f"Recall:    {rec_optimal:.4f}")
print(f"F1-Score:  {f1_optimal:.4f}")
print(f"ROC-AUC:   {roc_optimal:.4f}")
print(f"Confusion Matrix:\n{confusion_matrix(y_test, y_test_pred_optimal)}")

print(f"\nIMPROVEMENT WITH OPTIMAL THRESHOLD:")
print("-"*120)
print(f"Accuracy Improvement:  {acc_optimal - acc_default:+.4f}")
print(f"Precision Improvement: {prec_optimal - prec_default:+.4f}")
print(f"Recall Improvement:    {rec_optimal - rec_default:+.4f}")
print(f"F1-Score Improvement:  {f1_optimal - f1_default:+.4f}")

# ============ CONFIDENCE ANALYSIS ============
print(f"\n{'='*120}")
print(f"PREDICTION CONFIDENCE ANALYSIS")
print(f"{'='*120}")

correct_predictions = predictions_df[predictions_df['Prediction_Correct']]
incorrect_predictions = predictions_df[~predictions_df['Prediction_Correct']]

print(f"\nCorrect Predictions: {len(correct_predictions)} / {len(predictions_df)} ({len(correct_predictions)/len(predictions_df)*100:.2f}%)")
print(f"  Average Confidence: {correct_predictions['Confidence_Score'].mean():.4f}")
print(f"  Min Confidence:     {correct_predictions['Confidence_Score'].min():.4f}")
print(f"  Max Confidence:     {correct_predictions['Confidence_Score'].max():.4f}")

if len(incorrect_predictions) > 0:
    print(f"\nIncorrect Predictions: {len(incorrect_predictions)} / {len(predictions_df)} ({len(incorrect_predictions)/len(predictions_df)*100:.2f}%)")
    print(f"  Average Confidence: {incorrect_predictions['Confidence_Score'].mean():.4f}")
    print(f"  Min Confidence:     {incorrect_predictions['Confidence_Score'].min():.4f}")
    print(f"  Max Confidence:     {incorrect_predictions['Confidence_Score'].max():.4f}")

# ============ CLASS DISTRIBUTION IN PREDICTIONS ============
print(f"\n{'='*120}")
print(f"PREDICTION DISTRIBUTION")
print(f"{'='*120}")

print(f"\nActual Class Distribution:")
print(f"  Class 0 (No Stroke):    {(y_test == 0).sum()} ({(y_test == 0).sum()/len(y_test)*100:.2f}%)")
print(f"  Class 1 (Stroke):       {(y_test == 1).sum()} ({(y_test == 1).sum()/len(y_test)*100:.2f}%)")

print(f"\nPredicted Class Distribution (Default Threshold = 0.5):")
print(f"  Class 0 (No Stroke):    {(y_test_pred_default == 0).sum()} ({(y_test_pred_default == 0).sum()/len(y_test_pred_default)*100:.2f}%)")
print(f"  Class 1 (Stroke):       {(y_test_pred_default == 1).sum()} ({(y_test_pred_default == 1).sum()/len(y_test_pred_default)*100:.2f}%)")

print(f"\nPredicted Class Distribution (Optimal Threshold = {optimal_threshold:.4f}):")
print(f"  Class 0 (No Stroke):    {(y_test_pred_optimal == 0).sum()} ({(y_test_pred_optimal == 0).sum()/len(y_test_pred_optimal)*100:.2f}%)")
print(f"  Class 1 (Stroke):       {(y_test_pred_optimal == 1).sum()} ({(y_test_pred_optimal == 1).sum()/len(y_test_pred_optimal)*100:.2f}%)")
```
